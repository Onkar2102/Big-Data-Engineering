## 1. Introduction to Compression

Compression is a technique used to **reduce the size of data** stored in files, which helps in saving storage space and reducing I/O operations. In Hive, compression can be applied to both **row-based** and **column-based** file formats. However, the choice of compression algorithm depends on the trade-off between **compression ratio** and **compression speed**.

### Key Concepts:
- **Compression Ratio**: The ratio of the size of the uncompressed data to the size of the compressed data. Higher ratios mean better compression.
- **Compression Speed**: The time taken to compress and decompress data. Faster algorithms are preferred for real-time processing.
- **Splittability**: Whether the compressed file can be split into smaller parts for parallel processing.

### Why Use Compression?
- **Save Storage Space**: Reduces the amount of disk space required to store data.
- **Reduce I/O Operations**: Smaller files mean fewer I/O operations, improving query performance.
- **Improve Network Bandwidth**: Compressed data requires less bandwidth to transfer over the network.

## 2. Snappy Compression

### Overview:
- **Optimized for speed**: Provides moderate compression with very fast compression and decompression speeds.
- **Default compression** for Parquet and ORC file formats.
- **Splittable** when used with container-based file formats like ORC and Parquet.

### Advantages:
- **Fast compression and decompression**: Ideal for real-time processing.
- **Moderate compression ratio**: Balances speed and storage savings.

### Disadvantages:
- **Lower compression ratio**: Compared to Gzip or Bzip2, Snappy provides less compression.

### Example:
```sql
SET hive.exec.compress.output=true;
SET mapreduce.output.fileoutputformat.compress=true;
SET mapreduce.output.fileoutputformat.compress.codec=org.apache.hadoop.io.compress.SnappyCodec;

CREATE TABLE employees_snappy (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS ORC;
```

## 3. Gzip Compression

### Overview:
- **High compression ratio**: Provides excellent compression, reducing file sizes significantly.
- **Slower compression and decompression**: Not ideal for real-time processing.
- **Not splittable**: Cannot be divided into smaller parts for parallel processing.

### Advantages:
- **High compression ratio**: Ideal for archiving data where storage space is a concern.

### Disadvantages:
- **Slower compression and decompression**: Not suitable for real-time processing.
- **Not splittable**: Cannot be used for parallel processing in Hadoop.

### Example:
```sql
SET hive.exec.compress.output=true;
SET mapreduce.output.fileoutputformat.compress=true;
SET mapreduce.output.fileoutputformat.compress.codec=org.apache.hadoop.io.compress.GzipCodec;

CREATE TABLE employees_gzip (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS ORC;
```

## 4. Bzip2 Compression

### Overview:
- **Highest compression ratio**: Provides the best compression among all algorithms.
- **Very slow compression and decompression**: Not suitable for real-time processing.
- **Splittable**: Can be divided into smaller parts for parallel processing.

### Advantages:
- **Best compression ratio**: Ideal for long-term storage where space is critical.
- **Splittable**: Can be used for parallel processing in Hadoop.

### Disadvantages:
- **Very slow compression and decompression**: Not suitable for real-time processing.

### Example:
```sql
SET hive.exec.compress.output=true;
SET mapreduce.output.fileoutputformat.compress=true;
SET mapreduce.output.fileoutputformat.compress.codec=org.apache.hadoop.io.compress.BZip2Codec;

CREATE TABLE employees_bzip2 (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS ORC;
```

## 5. LZO Compression

### Overview:
- **Optimized for speed**: Provides fast compression and decompression.
- **Moderate compression ratio**: Balances speed and storage savings.
- **Splittable**: Can be divided into smaller parts for parallel processing.

### Advantages:
- **Fast compression and decompression**: Suitable for real-time processing.
- **Splittable**: Can be used for parallel processing in Hadoop.

### Disadvantages:
- **Requires a separate license**: Not included by default in Hadoop distributions.

### Example:
```sql
SET hive.exec.compress.output=true;
SET mapreduce.output.fileoutputformat.compress=true;
SET mapreduce.output.fileoutputformat.compress.codec=com.hadoop.compression.lzo.LzoCodec;

CREATE TABLE employees_lzo (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS ORC;
```

## 6. Zstandard Compression

### Overview:
- **High compression ratio**: Provides excellent compression, similar to Gzip.
- **Fast compression and decompression**: Comparable to Snappy in speed.
- **Splittable**: Can be divided into smaller parts for parallel processing.

### Advantages:
- **High compression ratio**: Ideal for both storage and real-time processing.
- **Fast compression and decompression**: Suitable for real-time processing.
- **Splittable**: Can be used for parallel processing in Hadoop.

### Disadvantages:
- **Less commonly used**: Not as widely supported as Snappy or Gzip.

### Example:
```sql
SET hive.exec.compress.output=true;
SET mapreduce.output.fileoutputformat.compress=true;
SET mapreduce.output.fileoutputformat.compress.codec=org.apache.hadoop.io.compress.ZStandardCodec;

CREATE TABLE employees_zstd (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS ORC;
```
