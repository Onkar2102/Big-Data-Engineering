# Upload the Data to Google bucket

```python
# dataproc-staging-us-central1-415554966392-gr7ktwvz/notebooks/jupyter/Project
# above is the path of the Project directory when accessed through Bucket portal

# BLOB = binary large object 

# pip install google google.cloud google-cloud-storage
```

```python
from google.cloud import storage
import os

# Define Variables
bucket_name = 'dataproc-staging-us-central1-415554966392-gr7ktwvz'
local_file_path = 'credit-card-approval-prediction/application_record.csv'
gcs_folder = 'notebooks/jupyter/Project/Data/'
gcs_file_path = gcs_folder + 'application_record.csv'

# # Initialize Google Cloud Storage Client
client = storage.Client()
bucket = client.bucket(bucket_name)

# Upload File
blob = bucket.blob(gcs_file_path)
blob.upload_from_filename(local_file_path)

print(f'File {local_file_path} uploaded to gs://{bucket_name}/{gcs_file_path}')
```

```python
# List files in the bucket
blobs = list(bucket.list_blobs(prefix=gcs_folder))
for blob in blobs:
    print(blob.name)
```
