## 1. Row-Based vs Column-Based Storage

### Row-Based Storage
In **row-based storage**, all the column values of a row are stored together, followed by the values of the next row. This format is ideal for **write-heavy workloads** where entire rows are frequently inserted or updated.

#### Example:
```
Row 1: [1, 'John', 'Doe', 50000]
Row 2: [2, 'Jane', 'Smith', 60000]
Row 3: [3, 'Alice', 'Johnson', 55000]
```

#### Advantages:
- **Efficient for writes**: Appending new rows is straightforward.
- **Good for transactional workloads**: Suitable for OLTP (Online Transaction Processing) systems.

#### Disadvantages:
- **Inefficient for reads**: Reading a subset of columns requires scanning the entire dataset.
- **Less compression**: Data of different types (e.g., integers and strings) are stored together, reducing compression efficiency.

### Column-Based Storage
In **column-based storage**, the values of a single column for all rows are stored together, followed by the values of the next column. This format is ideal for **read-heavy workloads** where only a subset of columns is frequently accessed.

#### Example:
```
Column 1 (ID): [1, 2, 3]
Column 2 (First Name): ['John', 'Jane', 'Alice']
Column 3 (Last Name): ['Doe', 'Smith', 'Johnson']
Column 4 (Salary): [50000, 60000, 55000]
```

#### Advantages:
- **Efficient for reads**: Only the relevant columns are read, reducing I/O.
- **Better compression**: Data of the same type is stored together, improving compression.

#### Disadvantages:
- **Inefficient for writes**: Updating or inserting new rows requires modifying multiple files.
- **Not ideal for transactional workloads**: More suited for OLAP (Online Analytical Processing) systems.

### Visualization:
```
Row-Based Storage:
[Row 1: 1, John, Doe, 50000]
[Row 2: 2, Jane, Smith, 60000]
[Row 3: 3, Alice, Johnson, 55000]

Column-Based Storage:
[Column 1: 1, 2, 3]
[Column 2: John, Jane, Alice]
[Column 3: Doe, Smith, Johnson]
[Column 4: 50000, 60000, 55000]
```

## 2. TextFile Format

### Overview:
- **Default file format** in Hive.
- Each line in the file is a record.
- Uses **LazySimpleSerDe** for serialization and deserialization.

### Advantages:
- **Human-readable**: Easy to view and edit.
- **Simple to use**: No special tools required.

### Disadvantages:
- **No compression**: Large file sizes.
- **No columnar storage**: Inefficient for queries accessing a subset of columns.

### Example:
```sql
CREATE TABLE employees_text (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
ROW FORMAT DELIMITED
FIELDS TERMINATED BY ','
STORED AS TEXTFILE;
```

## 3. SequenceFile Format

### Overview:
- **Binary key-value pairs** stored in a flat file.
- Used for intermediate data storage between MapReduce jobs.
- Uses **LazySimpleSerDe** for serialization and deserialization.

### Advantages:
- **Splittable**: Can be divided into smaller parts for parallel processing.
- **Supports compression**: Reduces storage space.

### Disadvantages:
- **Not human-readable**: Requires tools to view the data.
- **Less efficient for columnar queries**: Still row-based storage.

### Example:
```sql
CREATE TABLE employees_sequence (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS SEQUENCEFILE;
```

## 4. RCFile Format

### Overview:
- **Row Columnar File**: A hybrid format that stores data in a columnar format within row groups.
- Developed by Facebook for efficient storage and querying.
- Uses **ColumnarSerDe** for serialization and deserialization.

### Advantages:
- **Efficient for columnar queries**: Only relevant columns are read.
- **Supports compression**: Reduces storage space.

### Disadvantages:
- **Complex to use**: Requires understanding of row groups and sync markers.
- **Not as efficient as ORC or Parquet**: Less optimized for modern big data workloads.

### Example:
```sql
CREATE TABLE employees_rcfile (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS RCFILE;
```

## 5. ORCFile Format

### Overview:
- **Optimized Row Columnar**: A highly efficient columnar storage format.
- Designed for Hive, providing **efficient compression** and **fast query performance**.
- Uses **OrcSerde** for serialization and deserialization.

### Advantages:
- **High compression**: Reduces storage space significantly.
- **Efficient for columnar queries**: Only relevant columns are read.
- **Supports complex data types**: Structs, lists, maps, and unions.

### Disadvantages:
- **Not ideal for write-heavy workloads**: Better suited for read-heavy workloads.
- **Complex to use**: Requires understanding of stripes and indexes.

### Example:
```sql
CREATE TABLE employees_orc (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS ORC;
```

## 6. Parquet Format

### Overview:
- **Columnar storage format**: Designed for efficient storage and querying of nested data.
- Compatible with many big data tools like Spark, Impala, and Drill.
- Uses **ParquetHiveSerDe** for serialization and deserialization.

### Advantages:
- **Efficient for nested data**: Handles complex data types well.
- **High compression**: Reduces storage space.
- **Schema evolution**: Supports changes in schema over time.

### Disadvantages:
- **Not ideal for write-heavy workloads**: Better suited for read-heavy workloads.
- **Complex to use**: Requires understanding of column chunks and pages.

### Example:
```sql
CREATE TABLE employees_parquet (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS PARQUET;
```

## 7. Avro Format

### Overview:
- **Row-based storage format**: Ideal for write-heavy workloads.
- Supports **schema evolution**: Allows changes in schema over time.
- Uses **AvroSerDe** for serialization and deserialization.

### Advantages:
- **Efficient for writes**: Suitable for write-heavy workloads.
- **Schema evolution**: Supports changes in schema without breaking compatibility.
- **Splittable**: Can be divided into smaller parts for parallel processing.

### Disadvantages:
- **Less efficient for columnar queries**: Still row-based storage.
- **Complex to use**: Requires understanding of schema evolution.

### Example:
```sql
CREATE TABLE employees_avro (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS AVRO;
```

## 9. Code Examples

### Creating Tables with Different Formats
```sql
-- TextFile
CREATE TABLE employees_text (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
ROW FORMAT DELIMITED
FIELDS TERMINATED BY ','
STORED AS TEXTFILE;

-- SequenceFile
CREATE TABLE employees_sequence (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS SEQUENCEFILE;

-- ORCFile
CREATE TABLE employees_orc (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS ORC;

-- Parquet
CREATE TABLE employees_parquet (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS PARQUET;

-- Avro
CREATE TABLE employees_avro (
    employee_id INT,
    first_name STRING,
    last_name STRING,
    salary FLOAT
)
STORED AS AVRO;
```

### Loading Data into Tables
```sql
-- Load data into TextFile table
LOAD DATA INPATH '/path/to/data.txt' INTO TABLE employees_text;

-- Load data into ORCFile table
INSERT INTO TABLE employees_orc SELECT * FROM employees_text;

-- Load data into Parquet table
INSERT INTO TABLE employees_parquet SELECT * FROM employees_text;
```
